#Part 1 : Overall Statistical Analysis

```python
country_df = country_wise_latest.copy()
```

```python
statistics_table = pd.DataFrame({
    "Metric": ["Active", "Confirmed", "Deaths", "Recovered"],
    "Total": [
        country_df["Active"].sum(),
        country_df["Confirmed"].sum(),
        country_df["Deaths"].sum(),
        country_df["Recovered"].sum()
    ],
    "Average": [
        country_df["Active"].mean(),
        country_df["Confirmed"].mean(),
        country_df["Deaths"].mean(),
        country_df["Recovered"].mean()
    ],
    "Maximum": [
        country_df["Active"].max(),
        country_df["Confirmed"].max(),
        country_df["Deaths"].max(),
        country_df["Recovered"].max()
    ]
})

statistics_table
```


#Part 2 : Country-Level Analysis

`Top 10 Active`

```python
active_average = country_df["Active"].mean()

active_analysis = country_df[
    ["Country/Region", "Active"]
].copy()

active_analysis["Above Average"] = (
    active_analysis["Active"] > active_average
)

active_analysis = active_analysis.sort_values(
    "Active",
    ascending=False
)

active_analysis.head(10)
```

`Active_above_average`

```python
active_above_average = active_analysis[
    active_analysis["Above Average"] == True
]

active_above_average
```

`confirmed_above_average`

```python
confirmed_average = country_df["Confirmed"].mean()

confirmed_analysis = country_df[
    ["Country/Region", "Confirmed"]
].copy()

confirmed_analysis["Above Average"] = (
    confirmed_analysis["Confirmed"] > confirmed_average
)

confirmed_analysis = confirmed_analysis.sort_values(
    "Confirmed",
    ascending=False
)

confirmed_above_average = confirmed_analysis[
    confirmed_analysis["Above Average"] == True
]

confirmed_above_average
```

`deaths_above_average`

```python
deaths_average = country_df["Deaths"].mean()

deaths_analysis = country_df[
    ["Country/Region", "Deaths"]
].copy()

deaths_analysis["Above Average"] = (
    deaths_analysis["Deaths"] > deaths_average
)

deaths_analysis = deaths_analysis.sort_values(
    "Deaths",
    ascending=False
)

deaths_above_average = deaths_analysis[
    deaths_analysis["Above Average"] == True
]

deaths_above_average
```

`recovered_above_average`

```python
recovered_average = country_df["Recovered"].mean()

recovered_analysis = country_df[
    ["Country/Region", "Recovered"]
].copy()

recovered_analysis["Above Average"] = (
    recovered_analysis["Recovered"] > recovered_average
)

recovered_analysis = recovered_analysis.sort_values(
    "Recovered",
    ascending=False
)

recovered_above_average = recovered_analysis[
    recovered_analysis["Above Average"] == True
]

recovered_above_average
```

```python
print("Active:", len(active_above_average))
print("Confirmed:", len(confirmed_above_average))
print("Deaths:", len(deaths_above_average))
print("Recovered:", len(recovered_above_average))
```

# FINAL CLEAN DATASET

```python
final_data = full_grouped.copy()

# Convert Date to datetime
final_data["Date"] = pd.to_datetime(final_data["Date"])

# Sort data
final_data = final_data.sort_values(
    ["Country/Region", "Date"]
).reset_index(drop=True)

# Check final structure
print("Rows:", final_data.shape[0])
print("Columns:", final_data.shape[1])
print("Countries:", final_data["Country/Region"].nunique())
print("Dates:", final_data["Date"].nunique())

print("\nMissing values:")
print(final_data.isnull().sum())

print("\nDuplicate rows:")
print(final_data.duplicated().sum())
```
